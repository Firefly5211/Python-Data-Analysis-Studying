##### 创建s1，包含元素[92,67,70,88,76]及对应索引["小明","小红","小杰","小丽","小华"]。创建s2，包含元素[95,85,60,79,76]及对应索引["小明","小杰","小宇","小娟","小彤"]。创建后打印

```python
s1=pd.Series([92,67,70,88,76],index=["小明","小红","小杰","小丽","小华"])
s2=pd.Series([95,85,60,79,76],index=["小明","小杰","小宇","小娟","小彤"])
print(s1)
print(s2)
```

##### 让s1和s2的相同索引元素之间进行相加，如果无法找到对应索引值则将缺失值用0替代。

```python
s1.add(s2,fill_value=0)
```

```python
s1.describe()
```

##### 把s1更新为所有数字加5后的结果，并进行输出。

```python
s1=s1 + 5
```

##### 根据以下规则，输出一个新的Series，包含s1里各个学生成绩所对应的等级  
A：90以上  
B：80分及以上-90分  
C：70分及以上-80分  
D：60分及以上-70分  
不及格：60分及以下

```python
def oppo_grade_score(score):
    if score >=90:
        return "A"
    elif score >=80:
        return"B"
    elif score >=70:
        return"C"
    elif score>=60:
        return"D"
    else:
        return"不及格"
s1.apply(oppo_grade_score)
```
