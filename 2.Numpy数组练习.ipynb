```python
import numpy as a
```

```python
arr1=a.array([1,3,2,4,9])
arr2=a.ones(5)
arr3=a.zeros(3)
```

```python
arr4=a.concatenate([arr1,arr2,arr3])
arr4
```

#### 更改arr1 对arr1进行排序

```python
arr1.sort()
arr1
```

#### 用两种方法输出arr1中第三个元素

```python
print(arr1[2])
print(arr1[-3])
```

#### 输出从arr1第二个元素到第四个元素组成的数组

```python
arr1[1:4]
```

#### 输出arr1和arr2同位元素相加后得到的数组

```python
arr1 + arr2
```

#### 输出arr1各个元素乘以-5后组成的数组

```python
arr1 * -5
```

#### 输出arr1的最大值、最小值、总和、平均值 $(直接运行只会输出最后一行计算结果，要给每个式子加上print）$

```python
print(arr1.max())
print(arr1.min())
print(arr1.sum())
print(arr1.mean())
```

#### 用列表的形式能够把符合条件的数组元素取出，如果不加则会只输出布尔值（即：判断各元素是否符合括号内条件）

```python
arr1[(arr1 < 1) | (arr1 > 3)] 
```
