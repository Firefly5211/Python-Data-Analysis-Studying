# 项目：评估和清理英国电商公司销售数据

## 分析目标

此数据分析的目的是，根据市场销售数据，挖掘畅销产品，以便制定更有效的市场策略来提升营收。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了一家英国在线零售公司在2010年12月1日至2011年12月9日期间的所有交易情况，涵盖了该公司在全球不同国家和地区的业务数据。该公司主要销售覆盖各个场景的礼品，包括但不限于生日礼品、结婚纪念品、圣诞礼品等等。该公司的客户群体主要包括批发商和个人消费者，其中批发商占据了相当大的比例。

数据每列的含义如下：
- `InvoiceNo`: 发票号码。6位数，作为交易的唯一标识符。如果这个代码以字母“c”开头，表示这笔交易被取消。
- `StockCode`: 产品代码。5位数，作为产品的唯一标识符。
- `Description`: 产品名称。
- `Quantity`: 产品在交易中的数量。
- `InvoiceDate`: 发票日期和时间。交易发生的日期和时间。
- `UnitPrice`: 单价。价格单位为英镑（£）。
- `CustomerID`: 客户编号。5位数，作为客户的唯一标识符。
- `Country`: 国家名称。客户所居住的国家的名称。

## 二、评估数据

#### 1.对新建立的original_data这个DataFrame进行评估，主要是：

#### 2.DF的内容上存在一些问题：1.CustomerID和Description存在空缺值。   2.交易日期的数据类型是字符串   3.CustomerID是浮点数类型  
要进行数据类型转换、空缺值替换或删除

#### 评估缺失数据,了解哪些列存在空缺值后去筛选空缺

```python
original_data[original_data["Description"].isnull()]
```

##### 缺失Description的列Unitprice基本上为0,为了验证是否猜想正确，应当增加条件查看是否存在变量空缺且不为0的变量。

```python
original_data[(original_data["Description"].isnull()) & (original_data["UnitPrice"])!=0]
```

##### 通过筛选发现确实不存在变量不为0的变量，因此Description列空缺值对应的UnitPrice的确为0，因此这些观察值对目标没有价值，在后续清洗中应当删除。

```python
original_data[original_data["CustomerID"].isnull()]
```

#### 3.CustomerID、InvoiceNo、StockCode这几列数据根据现实情况分析是可以存在数据重复的，因此不对DF数据集中存在的重复数据进行处理。

#### 4.Country列可能存在不一致数据，例如国家简写不一致等。接下来要对该列进行检查

```python
original_data["Country"].value_counts()
```

##### 筛选后发现，美国(USA、United states)、英国(UK、U.K.)、澳大利亚(Australia、Austria)的指代值存在不一致现象，在清洗中应当保持一致。

#### 5.评估无效或错误的数据，通过describe方法快速了解数值统计信息。

```python
original_data.describe()
```

##### 通过大致了解可以发现Quantity、UnitPrice两列数值出现了问题，其值不应该为负数。  
可以先筛选出Quantity数值为负数的观察值，进一步评估其意义

```python
original_data[original_data["Quantity"]<0]
```

##### 通过筛选又发现StockCode列貌似出现了不符合标识规则的变量，且InvoiceNo中的变量疑似全部以c开头，即订单被取消。  
那么是否所有数量为负数的订单都被取消了？需要进一步对其进行筛选分析，看看是否有不以c开头且数量为负数的列。

```python
original_data[(original_data["Quantity"]<0) & (original_data["InvoiceNo"].str[0]!="C")]
```

##### 经过筛选发现确实存在不以c开头且数量为负数的列。猜测错误，同时发现：是否不以c开头且数量为负数的列的单价都是0？

```python
original_data[(original_data["Quantity"]<0) & (original_data["UnitPrice"]!= 0 ) & (original_data["InvoiceNo"].str[0]!="C")]
```

##### 因此当Quantity为负数时，要么订单被取消了，要么单价为0，这些都不是有效成交数据而且Quantity为负数还会影响分析，因此在清洗时应该将这些数据删除。接下来筛选UnitPrice为负数的行。

```python
original_data[original_data["UnitPrice"]<0]
```

##### 筛选发现，这两个观察值都是坏账调整(Adjust bad debt),不属于实际交易数据,可以在数据清洗时删除。

#### 6.清洗数据：  
1.InvoiceData变量的数据类型应该改成日期类；  
2.CustomerID变量的数据类型应该转换成字符串。  
3.Description变量的空缺值应当删除。  
4.美国(USA、United states)应该替换为“United States”
5.英国(UK、U.K.)应该替换为“United Kingdom”。  
6.Quantity为负数应该将这些数据删除。  
7.UnitPrice为负数删除。

##### 1.InvoiceData变量的数据类型应该改成日期类；

```python
cleaned_data=original_data.copy()
cleaned_data.head()
```

```python
cleaned_data["InvoiceDate"]=pd.to_datetime(cleaned_data["InvoiceDate"])
cleaned_data["InvoiceDate"]
```

##### 2.CustomerID变量的数据类型应该转换成字符串。

```python
cleaned_data["CustomerID"]=cleaned_data["CustomerID"].astype(str)
cleaned_data["CustomerID"]
```

##### 由于之前该列数据类型为float，因此转换类型后会有遗留的.0，接下来应该把.0进行删除操作进一步净化。

```python
cleaned_data["CustomerID"]=cleaned_data["CustomerID"].str.slice(0,-2)
cleaned_data
```

##### 3.Description变量的空缺值应当删除。还应当查看删除后该列空缺值的个数和。

```python
cleaned_data.dropna(subset=["Description"],inplace=True)
cleaned_data["Description"].isnull().sum()
```

##### 4.美国(USA、United states)应该替换为“United States”并检查替换后变量值为USA的个数。

```python
cleaned_data["Country"]=cleaned_data["Country"].replace({"USA":"United States"})
```

```python
len(cleaned_data[cleaned_data["Country"]=="USA"])
```

##### 5.英国(UK、U.K.)应该替换为“United Kingdom”。并检查替换后变量值为UK,U.K.的个数。

```python
cleaned_data["Country"]=cleaned_data["Country"].replace({"UK":"United Kingdom","U.K.":"United Kingdom"})
```

```python
print(len(cleaned_data[cleaned_data["Country"]=="UK"]))
print(len(cleaned_data[cleaned_data["Country"]=="U.K."]))
```

##### 6.Quantity为负数应该将这些数据删除,并检查替换后Quantity变量值为负数的个数。

```python
cleaned_data=cleaned_data[cleaned_data["Quantity"]>=0]
```

```python
len(cleaned_data[cleaned_data["Quantity"]<0])
```

##### 7.UnitPrice为负数删除。并检查替换后UnitPrice变量值为负数的个数。

```python
cleaned_data=cleaned_data[cleaned_data["UnitPrice"]>=0]
```

```python
len(cleaned_data[cleaned_data["UnitPrice"]<0])
```
