```python
import pandas as pd
import numpy as np
```

##### 创建一个名为students的DataFrame，包含数据如下：
{"001":{"姓名":"小陈","考试1":85,"考试2":92,"考试3":92}，"002":{"姓名":"小李","考试1":91,"考试2":92,"考试3":94}，"003":{"姓名":"小王","考试1":86,"考试2":81,"考试3":89}，"004":{"姓名":"小张","考试1":79,"考试2":89,"考试3":95}，"005":{"姓名":"小赵","考试1":96,"考试2":91,"考试3":91}，"006":{"姓名":"小周","考试1":81,"考试2":89,"考试3":92}}
要求索引为学号（001,、002等），列名为姓名、考试1、考试2、考试3，并输出创建好的DataFrame。

```python
student_data={"001":{"姓名":"小陈","考试1":85,"考试2":92,"考试3":92},"002":{"姓名":"小李","考试1":91,"考试2":92,"考试3":94},"003":{"姓名":"小王","考试1":86,"考试2":81,"考试3":89},"004":{"姓名":"小张","考试1":79,"考试2":89,"考试3":95},"005":{"姓名":"小赵","考试1":96,"考试2":91,"考试3":91},"006":{"姓名":"小周","考试1":81,"考试2":89,"考试3":92}}
students=pd.DataFrame(student_data).T
students
```

##### 创建一个叫student_average的DataFrame.索引为学号，包含姓名列，以及计算出三次考试平均值后的结果列(列名为"平均分")。

```python
average=students.loc[:,"考试1":"考试3"].mean(axis=1)
name=students["姓名"]
student_average=pd.DataFrame({"姓名":name,"平均分":average})
student_average
```

##### 输出每次考试第二高的分数组成的Series，可以尝试结合匿名函数lambda。

```python
grade=students.loc[:,"考试1":"考试3"].apply(lambda score: np.sort(score)[-2])
grade
```

##### 根据以下规则，输出每个学号对应的四次考试的等级：  
A+:平均分>=95  
A:90<=平均分<95  
B+:85<=平均分<90  
B:80<=平均分<85  
C+:75<=平均分<80  
C:平均分<75  


```python
def level(grade):
    if grade>=95:
        return "A+"
    elif 90<=grade<95:
        return "A"
    elif 85<=grade<90:
        return "B+"
    elif 80<=grade<85:
        return "B"
    elif 75<=grade<80:
        return "C+"
    else:
        return"C"
students.loc[:,"考试1":"考试3"].map(level)
```
